# 🔍 Indexing Libraries in Python for Vector Search

This notebook demonstrates how to use indexing libraries in Python for efficient vector similarity search.

We’ll cover:
- Flat (Brute-force) Index with FAISS
- Approximate Nearest Neighbor with FAISS: IVF, PQ, HNSW
- Optional: Spark-based indexing (simulation using Pandas if PySpark not installed)

```

import numpy as np
import pandas as pd
import faiss
from sklearn.metrics.pairwise import cosine_similarity
np.random.seed(42)
```

## 📄 Generate Synthetic Embeddings

```

# Generate 1000 random 128-d embeddings
dimension = 128
num_vectors = 1000
embeddings = np.random.rand(num_vectors, dimension).astype("float32")

# Create a random query
query = np.random.rand(1, dimension).astype("float32")
```

## 📦 Flat (Brute-force) Index

```

index_flat = faiss.IndexFlatL2(dimension)
index_flat.add(embeddings)
D_flat, I_flat = index_flat.search(query, k=5)

print("Top 5 results using Flat Index:")
print(I_flat)
```

## 📊 IVF (Inverted File) Index

```

nlist = 100  # number of clusters
quantizer = faiss.IndexFlatL2(dimension)
index_ivf = faiss.IndexIVFFlat(quantizer, dimension, nlist)

# Train the index with embeddings
index_ivf.train(embeddings)
index_ivf.add(embeddings)
D_ivf, I_ivf = index_ivf.search(query, k=5)

print("Top 5 results using IVF Index:")
print(I_ivf)
```

## 🧮 PQ (Product Quantization) Index

```

m = 16  # number of subquantizers
nbits = 8  # bits per code
index_pq = faiss.IndexPQ(dimension, m, nbits)
index_pq.train(embeddings)
index_pq.add(embeddings)
D_pq, I_pq = index_pq.search(query, k=5)

print("Top 5 results using PQ Index:")
print(I_pq)
```

## 🌐 HNSW (Graph-based) Index

```

index_hnsw = faiss.IndexHNSWFlat(dimension, 32)
index_hnsw.add(embeddings)
D_hnsw, I_hnsw = index_hnsw.search(query, k=5)

print("Top 5 results using HNSW Index:")
print(I_hnsw)
```

## ✅ Summary of Results

```

summary = pd.DataFrame({
    "Flat Index": I_flat[0],
    "IVF Index": I_ivf[0],
    "PQ Index": I_pq[0],
    "HNSW Index": I_hnsw[0]
})
summary
```
